# EDA (Análisis de Datos Exploratorio)

El análisis exploratorio (EDA) permite entender la estructura de los datos antes de modelar.
Se comprueban: valores nulos, tipos de datos, duplicados.

```python
Prestamos.dtypes
```

```python
Prestamos.info()
```

```python
Prestamos.describe()
```

```python
#filtrar el proposito por educacion
Prestamos_educacion = Prestamos[Prestamos["Proposito"] == "Educación"]
```

```python
Prestamos_educacion.head()
```

```python
Prestamos_educacion.duplicated().sum()
```

```python
#eliminar los duplicados de la base de datos Prestamos_educacion
Prestamos_educacion.drop_duplicates(inplace=True)
```

```python
Prestamos_educacion.isna().sum()
```

```python
Prestamos_educacion["Ratio_Interes"] = Prestamos_educacion["Ratio_Interes"].fillna(
    Prestamos_educacion["Ratio_Interes"].median())
```

```python
Prestamos_educacion["Estudios"] = Prestamos_educacion["Estudios"].fillna(
    Prestamos_educacion["Estudios"].mode()[0])
```
